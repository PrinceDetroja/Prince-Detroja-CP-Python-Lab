```python
#Write your own function (without using built in functions) to convert all characters of string in lower, upper and toggle case
def lower():
    str=input("Enter a string: ")
    for i in range(0,len(str)):
        if "a"<=str[i]<="z":
            print(str[i],end="")
        elif "A"<=str[i]<="Z":
            print(chr(ord(str[i])+32),end="")

def upper():
    str=input("Enter a string: ")
    for i in range(0,len(str)):
        if "A"<=str[i]<="Z":
            print(str[i],end="")
        elif "a"<=str[i]<="z":
            print(chr(ord(str[i])-32),end="")

def toggle():
    str=input("Enter a string: ")
    for i in range(0,len(str)):
        if "a"<=str[i]<="z":
            print(chr(ord(str[i])-32),end="")
        elif "A"<=str[i]<="Z":
            print(chr(ord(str[i])+32),end="")

lower()
upper()
toggle()
```
