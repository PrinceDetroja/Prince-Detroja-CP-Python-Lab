```python
#Generate N numbers of Fibonacci series(0,1,1,2,3)
def fibonacci():
    n=int(input("Enter number of terms: "))
    n1,n2=0,1
    count=0
    if n==0:
        print(n1)
    elif n==1:
        print(n2)
    else:
        while count<n:
            nth=n1+n2
            print(n1,end=" ")
            n1=n2
            n2=nth
            count+=1

fibonacci()
```
