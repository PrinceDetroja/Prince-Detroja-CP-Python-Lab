```python
#Check whether number is prime, perfect, armstrong, palindrome and automorphic
def prime(n):
    c=0
    for i in range(2,n):
        if n%i==0:
            c+=1

    if c==0:
        print(n,"is a prime number")
    else:
        print(n,"is not a prime number")
        

def perfect(n):
    c=0
    p=n/2
    for i in range(1,int(p+1)):
        if n%i==0:
            c+=i
    if c==n:
        print(n,"is a perfect number")
    else:
        print(n,"is not a perfect number")

def armstrong(n):
    c=0
    t=n
    while t>=1:
        c=c+(t%10)**3
        t=t//10
    if n==c:
        print(n,"is an armstrong number")
    else:
        print(n,"is not an armstrong number")

def palindrome(n):
    rev=0
    t=n
    while t>=1:
        dig=t%10
        rev=rev*10 + dig
        t=t//10
    if rev==n:
        print(n,"is a palindrome.")
    else:
        print(n,"is not a palindrome.")

def automorphic(n):
    x=n%10
    y=(n**2)%10
    if x==y:
        print(n,"is an automorphic number.")
    else:
        print(n,"is not an automorphic number.")


a=int(input("Enter a number: "))
prime(a)
perfect(a)
armstrong(a)
palindrome(a)
automorphic(a)
```
