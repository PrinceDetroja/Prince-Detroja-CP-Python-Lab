```python
#Print nCr and nPr
def pnc():
    n=int(input("Enter a positve number: "))
    r=int(input("Enter a positve number(less than n): "))
    x=1
    y=1
    z=1
    for i in range(0,n):
        x=x*(n-i)
    for i in range(0,r):
        y=y*(r-i)
    for i in range(0,n-r):
        z=z*(n-r-i)
    nCr=x/(y*z)
    nPr=x/z
    print(n,"C",r,"=",nCr)
    print(n,"P",r,"=",nPr)
pnc()
```
