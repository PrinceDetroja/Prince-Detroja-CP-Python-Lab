```python
"""
A list contains names of boys and girls as its elements. Boys name are stored as tuples. Write a program to find out no. of boys and girls  in the list
"""
def count():
    l=[("Prince","Aryan","Tirth","Devarya"),"Khushi","Heer",("Yug","Dhyaan"),"YashviM",("Rushi",)]
    nb=0
    ng=0
    for i in l:
        if isinstance(i, tuple):
            nb=nb+len(i)
        else:
            ng=ng+1
    print("No. of boys =",nb)
    print("No. of girls =",ng)

count()
```
