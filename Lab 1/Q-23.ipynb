```python
#Calculate average of 3 subjects along with their total
P=float(input('Enter marks in 1st subject out of 100: '))
Q=float(input('Enter marks in 2nd subject out of 100: '))
R=float(input('Enter marks in 3rd subject out of 100: '))
T=P+Q+R
A=T/3
print('Total marks obtained are=',T)
print('Average of 3 subjects is=',A)
```
