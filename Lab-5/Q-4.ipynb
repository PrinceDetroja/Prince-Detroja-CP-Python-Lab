```python
#Generate 30 random integers and store then in a list, Create 2 more list 1 with +ve numbers and 2 with -ve no.
import random
def separate():
    l1,l2,l3=[],[],[]
    for i in range(30):
        x=random.randint(-100,100)
        l1=l1+[x]
    print(l1)
    for i in l1:
        if i>=0:
            l2.append(i)
        else:
            l3.append(i)
    print("List with +ve no. is:",l2)
    print("List with -ve no. is:",l3)

separate()
```
