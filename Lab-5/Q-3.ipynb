```python
#Generate 50 random integers in range 1 and 30. Remove all duplicate values in list
import random
def remove():
    l1=[]
    for i in range(50):
        x=random.randint(1,30)
        l1=l1+[x]
    print(l1)
    l2=[]
    for i in l1:
        if i not in l2:
            l2.append(i)
    
    print(l2)

remove()
```
