```python
"""
Generate 20 random integers and store them in a list. Accept a number from a user and
print position of all occurrences of that number in list
"""
import random
def occurances():
    l1=[]
    for i in range(20):
        x=random.randint(0,100)
        l1=l1+[x]
    print(l1)
    n=int(input("Enter a number: "))
    print("Frequency of",n,"is:",l1.count(n))
    print("Postion of n in list:",end="")
    for i in range(len(l1)):
        if l1[i]==n:
            print(i,end=",")

occurances()
```
