```python
import random

odd_num = [random.randrange(1, 100, 2) for _ in range(5)]
print("Original odd numbers:", odd_num)

even_num = [random.randrange(0, 100, 2) for _ in range(4)]
print("Even numbers:", even_num)

odd_num[2] = even_num
print("After replacement:", odd_num)

flat_list = []
for item in odd_num:
    if isinstance(item, list):
        flat_list.extend(item)
    else:
        flat_list.append(item)
print("Flattened list:", flat_list)

flat_list.sort()
print("Final sorted list:", flat_list)
```
