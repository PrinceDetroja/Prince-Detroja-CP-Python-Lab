```python
#Take 2 list of no. and create 3rd list only for those no. which are in 1st list but not in 2nd(use list comprehension)
import random
def list():
    l1=[]
    l2=[]
    l3=[]
    for i in range(10):
        x=random.randint(-100,100)
        l1.append(x)
    for i in range(10):
        y=random.randint(-100,100)
        l2.append(y)
    print(l1)
    print(l2)
    for i in l1:
        if i not in l2:
            l3.append(i)
    print(l3)

list()
```
