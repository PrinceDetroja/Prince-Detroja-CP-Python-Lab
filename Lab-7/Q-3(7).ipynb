```python
def data():
    empdata = {(1,22):1000000, (1,10):50000, (2,9):300000}
    deptdata={}
    for k,v in empdata.items():
        print(k[0],k[1],v)
        if k[0] not in deptdata:
            deptdata[k[0]]={"Max":v, "Min":v, "Total":v}
        else:
            if v>deptdata[k[0]]["Max"]:
                deptdata[k[0]]["Max"]=v
            elif v<deptdata[k[0]]["Min"]:
                deptdata[k[0]]["Min"]=v
            deptdata[k[0]]["Total"] += v
    print(deptdata)

data()
```
